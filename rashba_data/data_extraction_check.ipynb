```python
import matplotlib.pyplot as plt
import pandas as pd
pd.set_option('display.max_columns', None)

nq = pd.read_pickle("no_que_combined.pkl")
nb = pd.read_pickle("no_bad_combined.pkl")
```

```python
fig, axs = plt.subplots(2,1, sharex = True)
# plt.subplot(2,1,1)
nb.hist('alpha', bins = 20, ax = axs[0], grid = False, rwidth = .85, edgecolor = 'k', color = 'blue', range = [0,8])
# plt.subplot(2,1,2)
nq.hist('alpha', bins = 20, ax = axs[1], grid = False, rwidth = .85, edgecolor = 'k', color = 'blue', range = [0,8])

axs[0].set_title("All bad removed")
axs[1].set_title("All questionable removed")
axs[0].semilogy()
axs[1].semilogy()
```
